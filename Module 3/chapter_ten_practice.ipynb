```python
def get_odds():
    for x in range (1, 10, 2):
        yield x

i = 1
for x in get_odds():
    if i == 3:
        print(x)
        break
    i += 1
```

```python
def test(func):
    def new(*args, **kwargs):
        print("start")
        result = func(*args, **kwargs)
        print("end")
        return result
    return new

@test
def old():
    print("Old function called, they said they want their outfit back.")

old()
```

```python
class OopsException(Exception):
    pass

raise OopsException()
```

```python
try:
    raise OopsException
except OopsException:
    print("Caught an oops")
```
